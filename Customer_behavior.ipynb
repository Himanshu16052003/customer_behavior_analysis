```python
import pandas as pd
```

```python
data =  pd.read_csv('customer_shopping_behavior.csv')
```

```python
data['Review Rating'] = data.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
data.isnull().sum()
```

```python
data.columns
```

```python
data.columns = data.columns.str.lower()
data.columns =  data.columns.str.replace(' ','_')
data  =  data.rename(columns = {'purchase_amount_(usd)': 'purchase_amount' } )
```

```python
data.columns
```

```python
# create a  columns  age_group
```

```python
labels  =['Young Adult', 'Adult', 'Middle-age' , 'Senior']
data['age_group'] =  pd.qcut(data['age'],q = 4, labels =  labels ) 
```

```python
data[['age', 'age_group']].head(10)
```

```python
#  Create column  Purchase_Frequency_days
```

```python
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' :  7,
    'Monthly' :  30,
    'Quarterly' :  90,
    'Bi- Weekly' :  14,
    'Annually' : 365,
    'Every 3 Months' :  90
}
data['purchase_frequency_days'] =  data['frequency_of_purchases'].map(frequency_mapping)
```

```python
data[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python

from sqlalchemy import create_engine
username = "postgres"
password = "Admin"
host = "localhost"
port = "5433"
database = "customer_behavior "

engine = create_engine(
    f"postgresql+psycopg://{username}:{password}@{host}:{port}/{database}"
)
table_name ="customer"
data.to_sql(table_name, engine, if_exists="replace", index=False)
print(f"Data successfully loaded into table '{table_name}' in data base '{database}'")
```
